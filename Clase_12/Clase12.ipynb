#### Ejercicio #4

Construir un script que reciba texto y lo convierta en vector a travez de algun embedding, mostrar espacio/graficar vectores embebidos

### Solución: embedding TF-IDF y visualización en 2D

TF-IDF representa cada texto como un vector: cada dimensión corresponde a una palabra del vocabulario y su valor mide qué tan relevante es esa palabra en el texto. PCA reduce esos vectores a dos dimensiones para poder graficarlos.

```python
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.metrics.pairwise import cosine_similarity

from embedding_utils import embed_texts

# El script recibe los textos en esta lista. Se pueden reemplazar por otros textos.
textos = [
    'La inteligencia artificial aprende patrones de los datos',
    'Los modelos de aprendizaje automático analizan datos',
    'El perro juega en el parque durante la tarde',
]

vectores, vectorizador = embed_texts(textos)
print('Vocabulario:', vectorizador.get_feature_names_out())
print('Matriz de embeddings (texto × palabra):')
print(vectores.toarray().round(3))

similitud = cosine_similarity(vectores)
print('\nSimilitud coseno entre textos:')
print(similitud.round(3))

# PCA proyecta cada embedding en un plano de dos dimensiones.
coordenadas = PCA(n_components=2, random_state=42).fit_transform(vectores.toarray())
plt.figure(figsize=(8, 5))
plt.scatter(coordenadas[:, 0], coordenadas[:, 1], s=120, color='royalblue')
for indice, texto in enumerate(textos):
    plt.annotate(f'Texto {indice + 1}', (coordenadas[indice, 0], coordenadas[indice, 1]),
                 xytext=(6, 6), textcoords='offset points')
plt.axhline(0, color='gray', linewidth=0.7)
plt.axvline(0, color='gray', linewidth=0.7)
plt.title('Textos representados en el espacio de embeddings (PCA)')
plt.xlabel('Componente principal 1')
plt.ylabel('Componente principal 2')
plt.grid(alpha=0.25)
plt.show()
```
